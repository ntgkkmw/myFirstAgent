# Codex MAS Quickstart
Interactively run the multi-agent workflow after installing dependencies.

```python
import json
from codex_mas.app.runner import Workflow, build_deps, get_settings
from codex_mas.schemas.models import UserQuery

settings = get_settings()
deps = build_deps(settings)
workflow = Workflow(deps)
query = UserQuery(goal='Summarize the architecture of codex mas')
result = workflow.run(query)
print(json.dumps(result.model_dump(), indent=2))
```
